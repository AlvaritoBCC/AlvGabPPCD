# Sesión de entornos en python: Anaconda, pip y venv

---

## 1️⃣ Crear el entorno conda con Python 3.11

```python

conda create --name mi_proyecto python=3.11 -y
```

---

## 2️⃣ Activar el entorno


```python
conda activate mi_proyecto
```

---

## 3️⃣ Instalar las librerías necesarias

```python
conda install numpy pandas matplotlib seaborn jupyter -y
```

---

## 4️⃣ Exportar dependencias

```python
pip freeze > requirements.txt
```

---
### Esto genera un archivo llamado `requirements.txt` con todas las versiones exactas de los paquetes instalados, por ejemplo:

### matplotlib==3.9.2
### numpy==2.1.1
### pandas==2.2.3
### seaborn==0.13.2
### jupyter==1.1.1

---
## Diferencias entre `conda`, `venv` y `pip`

| Herramienta | Descripción | Ventajas | Desventajas |
|--------------|--------------|-----------|--------------|
| **conda** | Gestor de entornos y paquetes multiplataforma (Python, R, C, etc.) | Administra entornos completos, resuelve conflictos de dependencias, permite usar distintas versiones de Python | Requiere instalar Anaconda/Miniconda, más pesado |
| **venv** | Módulo integrado en Python para crear entornos virtuales | Ligero, fácil de usar, no necesita instalar nada adicional | Solo maneja entornos Python (no librerías externas del sistema) |
| **pip** | Gestor de paquetes de Python que instala desde PyPI | Amplia compatibilidad, millones de paquetes disponibles | No crea entornos por sí solo (se usa junto con venv o conda) |
