```python
# Programa para calcular el IMC (Índice de Masa Corporal)

# Solicitamos los datos al usuario
# Nota: input() devuelve un string, por eso convertimos a float
peso = float(input("Ingresa tu peso en kilogramos (kg): "))
altura = float(input("Ingresa tu estatura en metros (m): "))

# Fórmula del IMC = peso / (altura ** 2)
imc = peso / (altura ** 2)

# Mostramos el resultado con 2 decimales
print("\nTu IMC es:", round(imc, 2))

# Clasificación según la OMS
if imc < 18.5:
    print("Estado: Bajo peso 🟡")
elif 18.5 <= imc < 25:
    print("Estado: Peso normal 🟢")
elif 25 <= imc < 30:
    print("Estado: Sobrepeso 🟠")
else:
    print("Estado: Obesidad 🔴")
```
