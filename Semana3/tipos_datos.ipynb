## Ejemplos base en código


```python
# Pedir número como string
numero_str = input("Digite un numero: ")
print("El numero digitado es: ", numero_str, "Cuyo tipo de datos es: ", type(numero_str))

# Convertir a entero
numero_int = int(numero_str)

# Sumar +3 en un ciclo de 3 veces
for i in range(0, 3):
    numero_int += 3
    print(numero_int)

# Convertir de nuevo a string
numero_str = str(numero_int)
print("El numero final es: ", numero_str, type(numero_str))
```

```python
# Solicitar un número al usuario
numero = input("Introduce un número: ")

# Mostrar el tipo original (str)
print("Tipo original:", type(numero))

# Convertir a entero
numero = int(numero)

# Mostrar el tipo después de la conversión
print("Tipo después de conversión:", type(numero))

# Usar un ciclo for para mostrar múltiplos
print("Múltiplos del número ingresado:")
for i in range(1, 6):
    print(numero * i)
```

```python
# Paso 1: Definimos un string
texto = "Hola"
print("Valor original:", texto, "| Tipo:", type(texto))

# Paso 2: Convertimos el string a booleano
# En Python, cualquier string NO vacío se considera True
valor_bool = bool(texto)
print("Convertido a booleano:", valor_bool, "| Tipo:", type(valor_bool))

# Paso 3: Convertimos el booleano a entero
# True se convierte en 1, False en 0
valor_int = int(valor_bool)
print("Convertido a entero:", valor_int, "| Tipo:", type(valor_int))
```
