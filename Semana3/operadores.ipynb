## Ejemplos base en código

```python
# Ejemplo 1: Calcular el volumen de un cilindro
# Fórmula: volumen = π * radio² * altura

import math

radio = 5
altura = 10
volumen = math.pi * (radio ** 2) * altura

print("1️⃣ Volumen del cilindro:", volumen)
```

```python
# Ejemplo 2: Verificar si un número está dentro de un rango
# Usamos operadores lógicos AND y de comparación

numero = 45
en_rango = numero >= 30 and numero <= 60

print("2️⃣ ¿El número está entre 30 y 60?:", en_rango)
```

```python
# Ejemplo 3: Determinar si un triángulo es equilátero, isósceles o escaleno

a, b, c = 5, 5, 8

if a == b == c:
    tipo = "Equilátero"
elif a == b or b == c or a == c:
    tipo = "Isósceles"
else:
    tipo = "Escaleno"

print("3️⃣ Tipo de triángulo:", tipo)
```

```python
# Ejemplo 4: Calcular descuento según monto de compra

compra = 850
descuento = 0.10 if compra > 500 else 0.05
total = compra - (compra * descuento)

print("4️⃣ Total a pagar con descuento:", total)
```

```python
# Ejemplo 5: Conversión de temperatura y evaluación lógica
# Convertir °C a °F y verificar si es un día caluroso (> 30°C o > 86°F)

celsius = 32
fahrenheit = (celsius * 9/5) + 32
caluroso = celsius > 30 or fahrenheit > 86

print("5️⃣ Temperatura:", celsius, "°C =", fahrenheit, "°F")
print("¿Día caluroso?:", caluroso)
```
