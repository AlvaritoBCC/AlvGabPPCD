En este notebook se presentan ejemplos básicos del uso de **Series** de Pandas,
mostrando distintas formas de creación y operaciones comunes.

## Importación de Pandas


```python
import pandas as pd
```

## Ejemplo 1: Serie a partir de una lista


```python
serie_lista = pd.Series([10, 20, 30, 40, 50])
serie_lista
```

## Ejemplo 2: Serie con índices personalizados


```python
serie_indices = pd.Series([100, 200, 300], index=["a", "b", "c"])
serie_indices
```

## Ejemplo 3: Serie a partir de un diccionario


```python
datos = {"manzana": 3, "banana": 5, "naranja": 2}
serie_diccionario = pd.Series(datos)
serie_diccionario
```

## Ejemplo 4: Operaciones básicas con Series


```python
# Operaciones matemáticas
serie_operaciones = serie_lista * 2
serie_operaciones
```

```python
# Operaciones estadísticas
print("Suma:", serie_lista.sum())
print("Media:", serie_lista.mean())
print("Máximo:", serie_lista.max())
print("Mínimo:", serie_lista.min())
```

## Ejemplo 5: Filtrado y selección de valores


```python
# Selección por condición
serie_filtrada = serie_lista[serie_lista > 25]
serie_filtrada
```
