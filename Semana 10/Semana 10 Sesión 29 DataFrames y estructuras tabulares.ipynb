## Importación de Pandas


```python
import pandas as pd
```

## Creación de un DataFrame con 10 registros


```python
# Crear un DataFrame con datos propios
datos = {
    "Nombre": ["Ana", "Luis", "Pedro", "María", "Carlos", "Lucía", "Jorge", "Sofía", "Miguel", "Elena"],
    "Edad": [25, 30, 28, 22, 35, 27, 40, 24, 33, 29],
    "Ciudad": ["Madrid", "Barcelona", "Sevilla", "Valencia", "Madrid",
               "Bilbao", "Zaragoza", "Granada", "Málaga", "Toledo"],
    "Salario": [1500, 1800, 1700, 1400, 2200, 1600, 2500, 1450, 2100, 1750]
}

df = pd.DataFrame(datos)
df
```

## Información básica del DataFrame

df.info()

df.describe()


## Selección de columnas y filas


```python
# Selección de una columna
df["Edad"]
# Selección de varias columnas
df[["Nombre", "Salario"]]
# Selección de filas por índice
df.loc[0:4]
```
