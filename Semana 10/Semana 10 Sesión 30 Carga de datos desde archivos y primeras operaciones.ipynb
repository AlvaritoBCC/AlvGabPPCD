y se realizan operaciones básicas de exploración y análisis inicial.


## Importación de librerías


```python
import pandas as pd
import numpy as np
```

## 1. Creación de un dataset realista y guardado en CSV


```python
# Fijar semilla para reproducibilidad
np.random.seed(10)

# Crear dataset con 60 registros
n = 60
datos = {
    "Nombre": [f"Empleado_{i+1}" for i in range(n)],
    "Edad": np.random.randint(20, 60, n),
    "Departamento": np.random.choice(
        ["Ventas", "IT", "RRHH", "Finanzas", "Marketing"], n
    ),
    "Salario": np.random.randint(1200, 3500, n)
}

df_original = pd.DataFrame(datos)

# Guardar dataset en CSV
df_original.to_csv("empleados.csv", index=False)
```

## 2. Carga del dataset desde el archivo CSV


```python
df = pd.read_csv("empleados.csv")
df.head()
```

## 3. Exploración inicial del dataset


```python
# Dimensiones del dataset
df.shape
# Información general
df.info()
# Estadísticas descriptivas
df.describe()
```

## 4. Inspección de columnas y datos


```python
# Nombres de columnas
df.columns
# Tipos de datos
df.dtypes
# Valores únicos por departamento
df["Departamento"].value_counts()
```

## 5. Operaciones básicas de análisis


```python
# Promedio de salario
df["Salario"].mean()
# Edad mínima y máxima
df["Edad"].min(), df["Edad"].max()
# Salario promedio por departamento
df.groupby("Departamento")["Salario"].mean()

```

## 6. Filtrado de datos


```python
# Empleados mayores de 40 años
df[df["Edad"] > 40].head()
# Empleados del departamento IT
df[df["Departamento"] == "IT"].head()

```
