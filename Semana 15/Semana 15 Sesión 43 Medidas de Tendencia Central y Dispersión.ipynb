En este notebook se realiza un análisis descriptivo utilizando medidas de tendencia
central y dispersión sobre distintos datasets reales, con el objetivo de comprender
la distribución y variabilidad de los datos.


## Importación de librerías


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Dataset 1: Titanic


```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

### Análisis descriptivo – Titanic

La edad presenta una dispersión moderada, mientras que la tarifa (`fare`) muestra
una alta variabilidad y posibles valores extremos, lo que sugiere una distribución
asimétrica.


```python
cols_titanic = ["age", "fare"]

titanic_stats = titanic[cols_titanic].dropna().agg(
    ["mean", "median", "var", "std", "min", "max"]
)

titanic_stats.loc["range"] = titanic_stats.loc["max"] - titanic_stats.loc["min"]
titanic_stats
```

## Dataset 2: Iris


```python
iris = sns.load_dataset("iris")
iris.head()
```

### Análisis descriptivo – Iris

Las medidas de tendencia central son muy cercanas entre sí, indicando distribuciones
bastante simétricas y con baja dispersión.


```python
cols_iris = ["sepal_length", "sepal_width"]

iris_stats = iris[cols_iris].agg(
    ["mean", "median", "var", "std", "min", "max"]
)

iris_stats.loc["range"] = iris_stats.loc["max"] - iris_stats.loc["min"]
iris_stats
```

## Dataset 3: Tips


```python
tips = sns.load_dataset("tips")
tips.head()
```

### Análisis descriptivo – Tips

El total de la cuenta presenta mayor dispersión que la propina. La diferencia entre
media y mediana sugiere una ligera asimetría en ambas variables.


```python
cols_tips = ["total_bill", "tip"]

tips_stats = tips[cols_tips].agg(
    ["mean", "median", "var", "std", "min", "max"]
)

tips_stats.loc["range"] = tips_stats.loc["max"] - tips_stats.loc["min"]
tips_stats
```

## Visualización opcional: Boxplots comparativos
Este gráfico refuerza visualmente la dispersión y presencia de posibles outliers.


```python
plt.figure()
sns.boxplot(data=tips[cols_tips])
plt.title("Boxplots de Total de Cuenta y Propina")
plt.show()
```
