En este notebook se identifican y analizan valores atípicos (outliers) utilizando
la regla del rango intercuartílico (IQR) sobre distintos datasets reales.


## Importación de librerías


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Dataset 1: Tips – Análisis de outliers en `total_bill`


```python
tips = sns.load_dataset("tips")
tips.head()
```

### Cálculo de límites de outliers (Regla 1.5 * IQR)


```python
variable = "total_bill"

Q1 = tips[variable].quantile(0.25)
Q3 = tips[variable].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

limite_inferior, limite_superior
```

### Boxplot e histograma


```python
plt.figure()
sns.boxplot(x=tips[variable])
plt.title("Boxplot del total de la cuenta")
plt.show()
```

```python
plt.figure()
sns.histplot(tips[variable], bins=20, kde=True)
plt.title("Histograma del total de la cuenta")
plt.show()
```

### Identificación de outliers

Los outliers pueden corresponder a cuentas inusualmente altas, posiblemente asociadas
a grupos grandes o eventos especiales. En un análisis real, estos valores podrían
mantenerse si representan casos válidos del negocio.


```python
outliers_tips = tips[
    (tips[variable] < limite_inferior) |
    (tips[variable] > limite_superior)
]

outliers_tips.shape
```

## Dataset 2: Titanic – Análisis de outliers en `fare`


```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

### Cálculo de límites de outliers (Regla 1.5 * IQR)


```python
variable = "fare"

Q1 = titanic[variable].quantile(0.25)
Q3 = titanic[variable].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

limite_inferior, limite_superior
```

### Boxplot e histograma


```python
plt.figure()
sns.boxplot(x=titanic[variable])
plt.title("Boxplot de la tarifa")
plt.show()
```

```python
plt.figure()
sns.histplot(titanic[variable], bins=30, kde=True)
plt.title("Histograma de la tarifa")
plt.show()
```

### Identificación de outliers

Las tarifas extremadamente altas pueden corresponder a pasajeros de primera clase.
Eliminar estos outliers podría distorsionar la realidad del dataset, por lo que
una transformación (como logaritmo) sería una mejor alternativa.


```python
outliers_titanic = titanic[
    (titanic[variable] < limite_inferior) |
    (titanic[variable] > limite_superior)
]

outliers_titanic.shape
```

## Opcional: Distribución sin outliers (Dataset Titanic)

Eliminar los outliers reduce la asimetría de la distribución, pero también elimina
información relevante sobre pasajeros con tarifas altas.


```python
titanic_sin_outliers = titanic[
    (titanic["fare"] >= limite_inferior) &
    (titanic["fare"] <= limite_superior)
]
```

```python
plt.figure()
sns.histplot(titanic_sin_outliers["fare"], bins=30, kde=True)
plt.title("Distribución de la tarifa sin outliers")
plt.show()
```
