## Importación de librerías


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Carga del dataset real

Se utiliza el dataset **tips**, que contiene información sobre cuentas y propinas
en un restaurante.


```python
df = sns.load_dataset("tips")
df.head()
```

## Exploración inicial del dataset


```python
df.info()
df.describe()
```

## Variables seleccionadas para el análisis

- Variables numéricas:
  - `total_bill`
  - `tip`
- Variable categórica:
  - `day`


## Estadística descriptiva

El total de la cuenta presenta mayor variabilidad que la propina. La diferencia entre
media y mediana sugiere una ligera asimetría positiva en ambas variables.


```python
cols_numericas = ["total_bill", "tip"]

stats = df[cols_numericas].agg(
    ["mean", "median", "var", "std", "min", "max"]
)

stats.loc["range"] = stats.loc["max"] - stats.loc["min"]
stats
```

## Análisis por variable categórica (día)
  
Existen diferencias en el gasto y las propinas promedio según el día de la semana.


```python
df.groupby("day")[cols_numericas].mean()
```

## Visualización 1: Histograma del total de la cuenta
La distribución muestra asimetría positiva, con algunas cuentas notablemente altas.


```python
plt.figure()
sns.histplot(df["total_bill"], bins=20, kde=True)
plt.title("Distribución del total de la cuenta")
plt.xlabel("Total de la cuenta")
plt.ylabel("Frecuencia")
plt.show()
```

## Visualización 2: Boxplot de la propina por día
Se observan diferencias en la dispersión de las propinas y la presencia de posibles
outliers en algunos días.


```python
plt.figure()
sns.boxplot(data=df, x="day", y="tip")
plt.title("Distribución de propinas por día")
plt.xlabel("Día")
plt.ylabel("Propina")
plt.show()
```

## Detección de outliers usando la regla 1.5 * IQR (total_bill)


```python
Q1 = df["total_bill"].quantile(0.25)
Q3 = df["total_bill"].quantile(0.75)
IQR = Q3 - Q1

limite_inf = Q1 - 1.5 * IQR
limite_sup = Q3 + 1.5 * IQR

limite_inf, limite_sup
```

### Registros identificados como outliers
 
Los outliers corresponden a cuentas inusualmente altas, probablemente asociadas a
grupos grandes o eventos especiales. No necesariamente deben eliminarse, ya que
representan casos reales del negocio.


```python
outliers = df[
    (df["total_bill"] < limite_inf) |
    (df["total_bill"] > limite_sup)
]

outliers.shape
```

## Visualización 3: Scatterplot – Total de la cuenta vs propina
Existe una relación positiva entre ambas variables: a mayor cuenta, mayor propina,
aunque con variabilidad según el día.


```python
plt.figure()
sns.scatterplot(data=df, x="total_bill", y="tip", hue="day")
plt.title("Relación entre total de la cuenta y propina")
plt.xlabel("Total de la cuenta")
plt.ylabel("Propina")
plt.show()
```
