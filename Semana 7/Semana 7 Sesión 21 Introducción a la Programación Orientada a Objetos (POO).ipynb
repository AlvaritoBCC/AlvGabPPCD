## Clase 1: Coche

La clase `Coche` representa un vehículo con atributos básicos y un método para arrancar.


```python
class Coche:
    def __init__(self, marca, modelo, anio):
        """
        Constructor de la clase Coche.
        
        Parámetros:
        marca (str): marca del coche
        modelo (str): modelo del coche
        anio (int): año del coche
        """
        self.marca = marca
        self.modelo = modelo
        self.anio = anio

    def arrancar(self):
        """
        Simula el arranque del coche.
        """
        return f"El coche {self.marca} {self.modelo} está arrancando."
```

```python
# Ejemplo de uso de la clase Coche
coche1 = Coche("Toyota", "Corolla", 2020)
print(coche1.arrancar())
```

## Clase 2: Libro

La clase `Libro` representa un libro con información básica y un método para mostrar su descripción.


```python
class Libro:
    def __init__(self, titulo, autor, paginas):
        """
        Constructor de la clase Libro.
        
        Parámetros:
        titulo (str): título del libro
        autor (str): autor del libro
        paginas (int): número de páginas
        """
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descripcion(self):
        """
        Devuelve una descripción del libro.
        """
        return f"'{self.titulo}' escrito por {self.autor}, {self.paginas} páginas."
```

```python
# Ejemplo de uso de la clase Libro
libro1 = Libro("Cien años de soledad", "Gabriel García Márquez", 471)
print(libro1.descripcion())
```

## Clase 3: Empleado

La clase `Empleado` representa a un empleado con su salario y permite aplicar un aumento porcentual.

```python
class Empleado:
    def __init__(self, nombre, salario):
        """
        Constructor de la clase Empleado.
        
        Parámetros:
        nombre (str): nombre del empleado
        salario (float): salario del empleado
        """
        self.nombre = nombre
        self.salario = salario

    def aumento(self, porcentaje):
        """
        Aplica un aumento al salario según un porcentaje dado.
        
        Parámetros:
        porcentaje (float): porcentaje de aumento
        """
        self.salario += self.salario * (porcentaje / 100)
        return self.salario
```

```python
# Ejemplo de uso de la clase Empleado
empleado1 = Empleado("Ana", 1500)
nuevo_salario = empleado1.aumento(10)
print(f"Nuevo salario de {empleado1.nombre}: {nuevo_salario}")
```
