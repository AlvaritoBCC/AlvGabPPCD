Este notebook muestra ejemplos de lectura y escritura de archivos `.txt` y `.csv`, el uso del bloque `with open()` y un ejercicio práctico que combina ambos tipos de archivos.

## 1. Lectura y escritura de archivos .txt

```python
# Escritura de un archivo .txt usando with open()
with open('ejemplo.txt', 'w', encoding='utf-8') as archivo:
    archivo.write('Nombre,Edad\n')
    archivo.write('Ana,25\n')
    archivo.write('Luis,30\n')
```

```python
# Lectura de un archivo .txt usando with open()
with open('ejemplo.txt', 'r', encoding='utf-8') as archivo:
    contenido = archivo.read()
    print(contenido)
```

## 2. Lectura y escritura de archivos .csv

```python
import csv

# Escritura de un archivo .csv
with open('ejemplo.csv', 'w', newline='', encoding='utf-8') as archivo_csv:
    writer = csv.writer(archivo_csv)
    writer.writerow(['Nombre', 'Edad'])
    writer.writerow(['Ana', 25])
    writer.writerow(['Luis', 30])
```

```python
# Lectura de un archivo .csv
with open('ejemplo.csv', 'r', encoding='utf-8') as archivo_csv:
    reader = csv.reader(archivo_csv)
    for fila in reader:
        print(fila)
```

## 3. Ejercicio práctico: de .txt a .csv

En este ejercicio se leen datos desde un archivo `.txt` y se guardan en un archivo `.csv`.

```python
# Crear un archivo .txt con datos
with open('datos.txt', 'w', encoding='utf-8') as archivo:
    archivo.write('Nombre,Edad\n')
    archivo.write('Pedro,28\n')
    archivo.write('María,22\n')
```

```python
# Leer el archivo .txt y guardarlo como .csv
with open('datos.txt', 'r', encoding='utf-8') as archivo_txt:
    lineas = archivo_txt.readlines()

with open('datos_convertidos.csv', 'w', newline='', encoding='utf-8') as archivo_csv:
    writer = csv.writer(archivo_csv)
    for linea in lineas:
        writer.writerow(linea.strip().split(','))
```

```python
# Verificar el contenido del nuevo archivo .csv
with open('datos_convertidos.csv', 'r', encoding='utf-8') as archivo_csv:
    reader = csv.reader(archivo_csv)
    for fila in reader:
        print(fila)
```
