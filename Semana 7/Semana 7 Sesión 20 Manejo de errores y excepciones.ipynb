## Programa 1: Conversión de tipos y validación numérica

Este programa solicita un valor al usuario e intenta convertirlo a número entero, manejando errores si la conversión no es válida.


```python
try:
    valor = input("Ingresa un número entero: ")
    numero = int(valor)
    print(f"Número ingresado correctamente: {numero}")
except ValueError:
    print("Error: el valor ingresado no es un número entero válido.")
else:
    print("Conversión realizada con éxito.")
finally:
    print("Fin del programa 1.\n")
```

## Programa 2: Lectura de archivos con control de errores

Este programa intenta leer un archivo de texto y muestra un mensaje adecuado si el archivo no existe.


```python
nombre_archivo = "archivo_ejemplo.txt"

try:
    with open(nombre_archivo, "r", encoding="utf-8") as archivo:
        contenido = archivo.read()
        print("Contenido del archivo:")
        print(contenido)
except FileNotFoundError:
    print(f"Error: el archivo '{nombre_archivo}' no existe.")
except IOError:
    print("Error: ocurrió un problema al leer el archivo.")
else:
    print("Archivo leído correctamente.")
finally:
    print("Fin del programa 2.\n")
```

## Programa 3: Excepción personalizada usando raise

En este programa se define una función que valida una edad.  
Si la edad es negativa, se lanza una excepción personalizada.


```python
def validar_edad(edad):
    """
    Valida que la edad sea un número positivo.
    
    Parámetros:
    edad (int): edad a validar.
    
    Lanza:
    ValueError: si la edad es negativa.
    """
    if edad < 0:
        raise ValueError("La edad no puede ser un número negativo.")
    return edad

try:
    edad_usuario = int(input("Ingresa tu edad: "))
    edad_valida = validar_edad(edad_usuario)
    print(f"Edad válida: {edad_valida}")
except ValueError as error:
    print(f"Error: {error}")
finally:
    print("Fin del programa 3.")
```
