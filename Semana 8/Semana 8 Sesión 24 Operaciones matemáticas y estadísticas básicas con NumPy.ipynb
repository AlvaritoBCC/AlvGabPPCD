## Importación de NumPy y semilla aleatoria


```python
import numpy as np

# Fijar semilla para resultados reproducibles
np.random.seed(42)
```

## 1. Conjuntos de datos aleatorios y estadísticas básicas


```python
# Conjunto 1: distribución uniforme (rand)
data1 = np.random.rand(10)

# Conjunto 2: distribución normal (randn)
data2 = np.random.randn(15)

# Conjunto 3: enteros aleatorios (randint)
data3 = np.random.randint(0, 100, 20)

# Conjunto 4: uniforme con más datos
data4 = np.random.rand(50)

# Conjunto 5: normal con más datos
data5 = np.random.randn(100)

datasets = [data1, data2, data3, data4, data5]
```

```python
# Cálculo de estadísticas
for i, data in enumerate(datasets, start=1):
    print(f"Dataset {i}")
    print("Media:", np.mean(data))
    print("Mediana:", np.median(data))
    print("Desviación estándar:", np.std(data))
    print("Mínimo:", np.min(data))
    print("Máximo:", np.max(data))
    print("-" * 30)
```

## 2. Matriz 5x5 y análisis por axis


```python
# Crear matriz 5x5 con enteros aleatorios
matriz = np.random.randint(1, 50, (5, 5))
matriz
```

```python
# Operaciones por filas (axis=1)
print("Suma por filas:", np.sum(matriz, axis=1))
print("Media por filas:", np.mean(matriz, axis=1))
print("Mínimo por filas:", np.min(matriz, axis=1))
print("Máximo por filas:", np.max(matriz, axis=1))
```

```python
# Operaciones por columnas (axis=0)
print("Suma por columnas:", np.sum(matriz, axis=0))
print("Media por columnas:", np.mean(matriz, axis=0))
print("Mínimo por columnas:", np.min(matriz, axis=0))
print("Máximo por columnas:", np.max(matriz, axis=0))
```

## 3. Simulación de 1,000 lanzamientos de un dado


```python
# Simulación de lanzamientos (valores del 1 al 6)
lanzamientos = np.random.randint(1, 7, 1000)

# Contar apariciones de cada cara
caras, conteos = np.unique(lanzamientos, return_counts=True)

# Calcular porcentajes
porcentajes = (conteos / len(lanzamientos)) * 100

for cara, porcentaje in zip(caras, porcentajes):
    print(f"Cara {cara}: {porcentaje:.2f}%")
```
