En este notebook se presentan ejemplos básicos del uso de la librería NumPy para la creación
y manipulación de arrays, comparándolos con listas de Python y explorando sus propiedades.

## Importación de NumPy


```python
import numpy as np
```

## 1. Creación de arrays con distintas funciones


```python
# Array a partir de una lista
array_lista = np.array([1, 2, 3, 4, 5])
array_lista
```

```python
# Array de ceros
array_ceros = np.zeros((2, 3))
array_ceros
```

```python
# Array de unos
array_unos = np.ones((3, 2))
array_unos
```

```python
# Array con valores consecutivos
array_arange = np.arange(0, 10, 2)
array_arange
```

```python
# Array con valores equiespaciados
array_linspace = np.linspace(0, 1, 5)
array_linspace
```

## 2. Comparación entre listas y arrays de NumPy


```python
# Lista de Python
lista = [1, 2, 3, 4, 5]

# Array de NumPy
array = np.array([1, 2, 3, 4, 5])

# Operaciones con listas
print("Lista + Lista:", lista + lista)

# Operaciones con arrays
print("Array + Array:", array + array)
```

```python
# Multiplicación
print("Lista * 2:", lista * 2)
print("Array * 2:", array * 2)
```

## 3. Propiedades de los arrays de NumPy


```python
array_ejemplo = np.array([[1, 2, 3],
                           [4, 5, 6]])

array_ejemplo
```

```python
# Forma del array
print("shape:", array_ejemplo.shape)

# Tamaño total (cantidad de elementos)
print("size:", array_ejemplo.size)

# Número de dimensiones
print("ndim:", array_ejemplo.ndim)

# Tipo de dato
print("dtype:", array_ejemplo.dtype)
```

## Conclusión

Los arrays de NumPy permiten realizar operaciones matemáticas de forma más eficiente
y sencilla que las listas de Python, además de ofrecer múltiples propiedades útiles
para el análisis de datos.
