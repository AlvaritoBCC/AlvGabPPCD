## Importación y creación del array base


```python
import numpy as np

# Array base para los ejemplos
array = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
array
```

## Ejemplo 1: Slicing básico


```python
# Selecciona los elementos desde la posición 2 hasta la 5 (sin incluir la 5)
slice_1 = array[2:5]

print("Array original:", array)
print("Slicing [2:5]:", slice_1)
```

## Ejemplo 2: Slicing con salto (step)


```python
# Selecciona elementos desde el inicio hasta el final, de dos en dos
slice_2 = array[::2]

print("Array original:", array)
print("Slicing [::2]:", slice_2)
```

## Ejemplo 3: Slicing inverso


```python
# Invierte el orden del array usando slicing
slice_3 = array[::-1]

print("Array original:", array)
print("Slicing inverso [::-1]:", slice_3)
```

## Ejemplo 4: Máscara booleana (valores mayores que 50)


```python
# Crea una máscara booleana
mask_mayores_50 = array > 50

print("Máscara booleana:", mask_mayores_50)
print("Valores mayores que 50:", array[mask_mayores_50])
```

## Ejemplo 5: Máscara booleana combinada


```python
# Selecciona valores mayores que 30 y menores o iguales a 80
mask_rango = (array > 30) & (array <= 80)

print("Máscara booleana:", mask_rango)
print("Valores entre 30 y 80:", array[mask_rango])
```

## Comparación visual de resultados


```python
print("Array original:", array)
print("Slicing básico:", slice_1)
print("Slicing con salto:", slice_2)
print("Slicing inverso:", slice_3)
print("Máscara > 50:", array[mask_mayores_50])
print("Máscara 30 < x <= 80:", array[mask_rango])
```
