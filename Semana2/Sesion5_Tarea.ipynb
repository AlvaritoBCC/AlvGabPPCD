### Diferencias entre Jupyter Notebook y Google Colab

| Característica | **Jupyter Notebook** | **Google Colab** |
|----------------|----------------------|------------------|
| Ejecución | Local (en tu computadora) | En la nube (servidores de Google) |
| Instalación | Necesita instalar Anaconda o Jupyter | No requiere instalación |
| Almacenamiento | Archivos guardados localmente | Guarda en Google Drive |
| Rendimiento | Depende de tu equipo | Puede usar GPU/TPU gratuitas |
| Colaboración | Limitada (requiere Git o compartir archivos) | Permite colaboración en tiempo real |

```python
import pandas as pd

# Un ejemplo de Python con Pandas cargando un dataset pequeño (CSV).
data = {
    "Nombre": ["Ana", "Luis", "María", "Carlos", "Sofía"],
    "Edad": [23, 35, 29, 40, 31],
    "Puntaje": [85, 90, 88, 95, 92]
}

df = pd.DataFrame(data)
df.to_csv("datos.csv", index=False)

# Cargar el CSV
df_cargado = pd.read_csv("datos.csv")

# Mostrar el DataFrame
print("📄 Dataset cargado:")
display(df_cargado)
```

-----

## Un grafico con Matplotlib

-----

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.barplot(data=df_cargado, x="Nombre", y="Puntaje", palette="coolwarm")
plt.title("Puntaje por Persona", fontsize=14)
plt.xlabel("Nombre")
plt.ylabel("Puntaje")
plt.tight_layout()
plt.show()
```

```python
# Importar la librería
import matplotlib.pyplot as plt

# Datos
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]
ventas = [120, 150, 180, 90, 200, 230]

#gráfico
plt.figure(figsize=(8, 5))  # Tamaño del gráfico
plt.plot(meses, ventas, marker='o', linestyle='-', color='blue', label='Ventas')


plt.title("Ventas Mensuales", fontsize=16)
plt.xlabel("Mes", fontsize=12)
plt.ylabel("Cantidad Vendida", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Mostrar el gráfico
plt.show()
```
