## Ventajas de usar Visual Studio Code (VS Code)

1. Integración con entornos virtuales (Conda o venv) - Permite seleccionar fácilmente el intérprete de Python asociado a tu entorno.  
2. Compatibilidad con Jupyter Notebooks - Puedes ejecutar y visualizar celdas directamente dentro del editor.  
3. Control de versiones con Git - Integración nativa con Git y GitHub para subir, clonar o hacer commits desde la interfaz.  
4. Extensiones potentes - Autocompletado, depuración, ejecución de código y herramientas de análisis estático.  
5. Ligero y multiplataforma - Funciona en Windows, macOS y Linux con excelente rendimiento.


```python

import pandas as pd

# Crear un pequeño DataFrame de ejemplo
data = {
    "Nombre": ["Ana", "Luis", "María", "Carlos", "Sofía"],
    "Edad": [23, 35, 29, 40, 31],
    "Puntaje": [85, 90, 88, 95, 92]
}

df = pd.DataFrame(data)
df.to_csv("datos.csv", index=False)

# Leer el archivo CSV
df_leido = pd.read_csv("datos.csv")
print("Datos cargados correctamente:")
display(df_leido)
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.barplot(data=df_leido, x="Nombre", y="Puntaje", palette="viridis")
plt.title("Puntaje por Persona", fontsize=14)
plt.xlabel("Nombre")
plt.ylabel("Puntaje")
plt.tight_layout()
plt.show()
```
