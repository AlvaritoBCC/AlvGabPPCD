## Importación de NumPy y configuración inicial


```python
import numpy as np

# Fijar semilla para reproducibilidad
np.random.seed(123)
```

## 1. Análisis de un dataset aleatorio 10x10


```python
# Crear dataset 10x10 con distribución normal
data = np.random.randn(10, 10)
data
```

```python
# Estadísticas globales
print("Media global:", np.mean(data))
print("Varianza global:", np.var(data))
print("Desviación estándar global:", np.std(data))
print("Mínimo global:", np.min(data))
print("Máximo global:", np.max(data))
```

```python
# Estadísticas por filas (axis=1)
print("\n--- Estadísticas por filas ---")
print("Media:", np.mean(data, axis=1))
print("Varianza:", np.var(data, axis=1))
print("Desviación estándar:", np.std(data, axis=1))
print("Mínimo:", np.min(data, axis=1))
print("Máximo:", np.max(data, axis=1))
```

```python
# Estadísticas por columnas (axis=0)
print("\n--- Estadísticas por columnas ---")
print("Media:", np.mean(data, axis=0))
print("Varianza:", np.var(data, axis=0))
print("Desviación estándar:", np.std(data, axis=0))
print("Mínimo:", np.min(data, axis=0))
print("Máximo:", np.max(data, axis=0))
```

## 2. Percentiles e interpretación


```python
# Cálculo de percentiles
percentiles = [10, 25, 50, 75, 90]
valores_percentiles = np.percentile(data, percentiles)

for p, v in zip(percentiles, valores_percentiles):
    print(f"Percentil {p}: {v}")
```

## 3. Experimento de correlación entre variables


```python
# Generar variables
A = np.random.randn(100)
B = 2 * A + np.random.randn(100) * 0.2  # Dependencia lineal con ruido leve
C = np.random.randn(100)               # Ruido independiente

# Matriz de datos
variables = np.vstack([A, B, C])

# Cálculo de la matriz de correlación
correlacion = np.corrcoef(variables)
correlacion
```
