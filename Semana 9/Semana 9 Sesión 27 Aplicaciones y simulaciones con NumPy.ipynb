En este notebook se presenta una simulación de un experimento aleatorio utilizando NumPy,
acompañada de análisis estadístico básico, gráficas simples y una breve interpretación
de los resultados obtenidos.

## Importación de librerías y configuración inicial


```python
import numpy as np
import matplotlib.pyplot as plt

# Semilla para reproducibilidad
np.random.seed(42)
```

## 1. Simulación de un experimento aleatorio: Lanzamiento de un dado

Se simulan 10,000 lanzamientos de un dado justo (caras del 1 al 6).


```python
# Simulación de lanzamientos
lanzamientos = np.random.randint(1, 7, 10000)

lanzamientos[:10]  # vista previa
```

## 2. Análisis estadístico del experimento


```python
# Medidas de tendencia central
media = np.mean(lanzamientos)
mediana = np.median(lanzamientos)

# Medidas de dispersión
std = np.std(lanzamientos)
minimo = np.min(lanzamientos)
maximo = np.max(lanzamientos)

print("Media:", media)
print("Mediana:", mediana)
print("Desviación estándar:", std)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
# Frecuencia de cada cara
caras, conteos = np.unique(lanzamientos, return_counts=True)
porcentajes = conteos / len(lanzamientos) * 100

for cara, porcentaje in zip(caras, porcentajes):
    print(f"Cara {cara}: {porcentaje:.2f}%")
```

## 3. Visualización de resultados


```python
# Gráfica de barras de frecuencias
plt.figure()
plt.bar(caras, conteos)
plt.xlabel("Cara del dado")
plt.ylabel("Frecuencia")
plt.title("Frecuencia de aparición de cada cara del dado")
plt.show()
```

```python
# Histograma de los lanzamientos
plt.figure()
plt.hist(lanzamientos, bins=6)
plt.xlabel("Valor del dado")
plt.ylabel("Cantidad")
plt.title("Histograma de lanzamientos del dado")
plt.show()
```
