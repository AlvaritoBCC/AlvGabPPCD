# Mini-Proyecto: Registro de Estudiantes

## Funciones del Sistema

- agregar_estudiante(registro, estudiante) → agrega un nuevo estudiante validando los campos.
- listar_estudiantes(registro) → imprime una tabla con todos los estudiantes.
- buscar_por_carrera(registro, carrera) → devuelve una lista filtrada por carrera.
- mejor_promedio(registro) → devuelve (nombre, promedio) del estudiante con la nota más alta.
- (Opcional) Calcular el promedio general.

```python
from typing import List, Dict, Any, Tuple

# Alias de tipo para mayor legibilidad
Registro = List[Dict[str, Any]]

def agregar_estudiante(registro: Registro, estudiante: Dict[str, Any]) -> None:
    """
     Agrega un nuevo estudiante al registro validando campos básicos.
    
    Validaciones:
    - nombre: str no vacío
    - edad: int > 0
    - carrera: str no vacío
    - promedio: número entre 0 y 10
    """
    assert isinstance(estudiante.get("nombre"), str) and estudiante["nombre"], "Nombre inválido"
    assert isinstance(estudiante.get("edad"), int) and estudiante["edad"] > 0, "Edad inválida"
    assert isinstance(estudiante.get("carrera"), str) and estudiante["carrera"], "Carrera inválida"
    assert isinstance(estudiante.get("promedio"), (int, float)) and 0 <= estudiante["promedio"] <= 10, "Promedio inválido"
    registro.append(estudiante)


def listar_estudiantes(registro: Registro) -> None:
    """
     Imprime una tabla simple con los estudiantes del registro.
    """
    if not registro:
        print(" No hay estudiantes registrados.")
        return
    
    print(f"{'Nombre':<12} {'Edad':<4} {'Carrera':<12} {'Promedio':<8}")
    print("-"*42)
    for est in registro:
        print(f"{est['nombre']:<12} {est['edad']:<4} {est['carrera']:<12} {est['promedio']:<8.2f}")


def buscar_por_carrera(registro: Registro, carrera: str) -> Registro:
    """
     Devuelve una lista filtrada de estudiantes de una carrera específica.
    """
    return [e for e in registro if e.get("carrera", "").lower() == carrera.lower()]


def mejor_promedio(registro: Registro) -> Tuple[str, float]:
    """
     Devuelve el (nombre, promedio) del estudiante con la nota más alta.
    Si el registro está vacío, devuelve ("", 0.0).
    """
    if not registro:
        return ("", 0.0)
    m = max(registro, key=lambda e: e.get("promedio", 0))
    return (m["nombre"], m["promedio"])


def promedio_general(registro: Registro) -> float:
    """
     Calcula el promedio general de todos los estudiantes registrados.
    """
    if not registro:
        return 0.0
    return sum(e["promedio"] for e in registro) / len(registro)
```

```python
# Inicializar registro vacío
registro: Registro = []

# Agregar estudiantes
agregar_estudiante(registro, {"nombre": "Ana", "edad": 20, "carrera": "Datos", "promedio": 9.1})
agregar_estudiante(registro, {"nombre": "Luis", "edad": 22, "carrera": "IA", "promedio": 8.7})
agregar_estudiante(registro, {"nombre": "Marta", "edad": 21, "carrera": "Datos", "promedio": 9.5})

# Listar todos los estudiantes
print(" Registro completo:\n")
listar_estudiantes(registro)

# Buscar estudiantes por carrera
print("\n Estudiantes de la carrera 'Datos':")
listar_estudiantes(buscar_por_carrera(registro, "Datos"))

# Mostrar el mejor promedio
nombre, promedio = mejor_promedio(registro)
print(f"\n Mejor promedio: {nombre} ({promedio:.2f})")

# Promedio general
print(f"\n Promedio general del grupo: {promedio_general(registro):.2f}")
```

# Estructuras usadas:


| Función                | Qué hace                                  | Estructura usada                |
| ---------------------- | ----------------------------------------- | ------------------------------- |
| `agregar_estudiante()` | Valida y agrega un nuevo registro         | Diccionario (clave:valor)       |
| `listar_estudiantes()` | Recorre e imprime todos los registros     | Lista de diccionarios           |
| `buscar_por_carrera()` | Filtra la lista por coincidencia de texto | Comprensión de listas           |
| `mejor_promedio()`     | Busca el valor máximo basado en una clave | Función `max()`                 |
| `promedio_general()`   | Calcula el promedio global                | `sum()` y comprensión de listas |
