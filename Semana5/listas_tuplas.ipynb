# Contenidos
## Creación de listas y tuplas
- mi_lista = [1, 2, 3]
- mi_tupla = (1, 2, 3)

## Indexación y slicing
- print(mi_lista[0])   # primer elemento
- print(mi_lista[-1])  # último elemento
- print(mi_lista[1:3]) # elementos desde el índice 1 hasta 2

## Métodos de listas más comunes
- append() -> agrega un elemento al final
- insert() -> inserta en una posición específica
- remove() -> elimina un elemento por valor
- pop() -> elimina un elemento por índice
- sort() -> ordena la lista
- reverse() -> invierte el orden

## Tuplas
- Inmutables: no se pueden modificar después de crearse.
- Útiles para datos que no deben cambiar (ej. coordenadas).

```python
# Registro de Calificaciones

calificaciones = []

n = int(input(" ¿Cuántos estudiantes hay? "))

for i in range(n):
    nota = float(input(f"Ingrese la nota del estudiante {i+1}: "))
    calificaciones.append(nota)

promedio = sum(calificaciones) / len(calificaciones)
print(f"\n Promedio general: {promedio:.2f}")
print(f"🔺 Nota más alta: {max(calificaciones)}")
print(f"🔻 Nota más baja: {min(calificaciones)}")
```

```python
# Lista de Compras

lista_compras = []

while True:
    accion = input("\n Escribe 'agregar', 'eliminar' o 'salir': ").lower()

    if accion == "salir":
        break
    elif accion == "agregar":
        producto = input(" Producto a agregar: ")
        lista_compras.append(producto)
    elif accion == "eliminar":
        producto = input(" Producto a eliminar: ")
        if producto in lista_compras:
            lista_compras.remove(producto)
        else:
            print(" No está en la lista.")
    else:
        print(" Acción no reconocida.")

print(f"\n Lista final: {lista_compras}")
```

```python
import random

#  Sorteo Aleatorio de Equipos

jugadores = ["Gab", "Anel", "Jorge", "Zárate", "Mina", "Leo"]
random.shuffle(jugadores)

equipos = []
for i in range(0, len(jugadores), 2):
    equipo = (jugadores[i], jugadores[i+1])
    equipos.append(equipo)

print("\n Equipos formados:")
for e in equipos:
    print(f"- {e[0]} y {e[1]}")
```

```python
#  Inventario con Búsqueda

inventario = [
    ("Teclado", 5, 25.0),
    ("Mouse", 10, 15.0),
    ("Monitor", 3, 150.0)
]

producto = input(" Ingresa el nombre del producto a buscar: ").capitalize()

for item in inventario:
    nombre, cantidad, precio = item
    if nombre == producto:
        print(f"{nombre}: {cantidad} unidades, total = ${cantidad * precio}")
        break
else:
    print(" Producto no encontrado.")
```

```python
#  Promedio y Ranking de Puntajes

puntajes = [
    ("Gab", 95),
    ("Anel", 82),
    ("Capitano", 89),
    ("Zárate", 93),
    ("Mina", 85)
]

# Ordenar por puntaje (índice 1) de mayor a menor
puntajes_ordenados = sorted(puntajes, key=lambda x: x[1], reverse=True)

print(" Ranking de Puntajes:")
for i, (nombre, score) in enumerate(puntajes_ordenados[:3], start=1):
    print(f"{i}. {nombre} — {score} puntos")

promedio = sum(p[1] for p in puntajes) / len(puntajes)
print(f"\n Promedio general: {promedio:.2f}")
```
