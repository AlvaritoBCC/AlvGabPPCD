```python
def agenda_contactos():
    """
     Agenda interactiva de contactos
    Permite agregar, buscar, actualizar y eliminar contactos usando un diccionario.
    """
    agenda = {}

    while True:
        print("\n Opciones: [1] Agregar | [2] Buscar | [3] Actualizar | [4] Eliminar | [5] Salir")
        opcion = input(" Elige una opción: ")

        if opcion == "1":
            nombre = input("Nombre: ").capitalize()
            telefono = input("Teléfono: ")
            agenda[nombre] = telefono
            print(" Contacto agregado.")
        elif opcion == "2":
            nombre = input("Nombre a buscar: ").capitalize()
            print(f" {nombre}: {agenda.get(nombre, 'No encontrado')}")
        elif opcion == "3":
            nombre = input("Nombre a actualizar: ").capitalize()
            if nombre in agenda:
                nuevo = input("Nuevo teléfono: ")
                agenda[nombre] = nuevo
                print(" Contacto actualizado.")
            else:
                print(" No existe ese contacto.")
        elif opcion == "4":
            nombre = input("Nombre a eliminar: ").capitalize()
            if nombre in agenda:
                del agenda[nombre]
                print(" Contacto eliminado.")
            else:
                print(" No existe ese contacto.")
        elif opcion == "5":
            break
        else:
            print(" Opción inválida.")
```

```python
def frecuencia_palabras(texto):
    """
     Cuenta la frecuencia de cada palabra en un texto.
    Devuelve un diccionario con conteos y muestra las 3 más frecuentes.
    """
    palabras = texto.lower().split()
    conteo = {}

    for palabra in palabras:
        conteo[palabra] = conteo.get(palabra, 0) + 1

    # Ordenar por frecuencia descendente
    top3 = sorted(conteo.items(), key=lambda x: x[1], reverse=True)[:3]

    print(" Top 3 palabras más comunes:")
    for palabra, frecuencia in top3:
        print(f"{palabra}: {frecuencia}")

    return conteo

#  Ejemplo de uso
texto = "Python es poderoso y divertido. Python es fácil de aprender y poderoso."
frecuencia_palabras(texto)
```

```python
def comparar_materias():
    """
     Compara alumnos inscritos en dos materias usando conjuntos (set).
    Muestra intersección, unión y diferencia.
    """
    matematicas = {"Gab", "Anel", "Mina", "Zárate"}
    programacion = {"Capitano", "Gab", "Leo", "Mina"}

    print(" Alumnos en Matemáticas:", matematicas)
    print(" Alumnos en Programación:", programacion)

    print("\n En ambas materias:", matematicas & programacion)
    print(" En al menos una:", matematicas | programacion)
    print(" Solo en Matemáticas:", matematicas - programacion)
```

```python
def nombres_unicos(nombres):
    """
     Recibe una lista de nombres y devuelve el conjunto de nombres únicos.
    También muestra cuántos nombres distintos hay.
    """
    unicos = set(nombres)
    print(f" Nombres únicos: {unicos}")
    print(f" Total distintos: {len(unicos)}")

#  Ejemplo de uso
nombres_unicos(["Gab", "Anel", "Gab", "Mina", "Capitano", "Mina"])
```

```python
def validar_correos(lista_correos):
    """
     Valida una lista de correos electrónicos.
    Usa un set para detectar duplicados y un dict para marcar válidos o inválidos.
    """
    correos_vistos = set()
    resultados = {}

    for correo in lista_correos:
        if correo in correos_vistos:
            resultados[correo] = " Duplicado"
        elif "@" in correo and "." in correo:
            resultados[correo] = " Válido"
        else:
            resultados[correo] = " Inválido"
        correos_vistos.add(correo)

    return resultados

#  Ejemplo de uso
correos = [
    "gab@gmail.com",
    "anelhotmail.com",
    "mina@yahoo.com",
    "gab@gmail.com",
    "leo@outlook"
]

resultado = validar_correos(correos)
for correo, estado in resultado.items():
    print(f"{correo}: {estado}")
```

----
| Concepto | Estructura               | Propósito                           |                                  |
| -------- | ------------------------ | ----------------------------------- | -------------------------------- |
| `dict`   | `{clave: valor}`         | Asociar datos (clave → valor)       |                                  |
| `set`    | `{elemento1, elemento2}` | Almacenar valores únicos            |                                  |
| `.get()` | Diccionario              | Evita errores si la clave no existe |                                  |
| `&`, `   | `, `-`                   | Conjuntos                           | Intersección, unión y diferencia |
