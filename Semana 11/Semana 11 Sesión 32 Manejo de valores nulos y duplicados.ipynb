En este notebook se trabajan técnicas básicas de limpieza de datos reales,
incluyendo la identificación de valores nulos, distintas estrategias de imputación
y la detección y eliminación de registros duplicados.

## Importación de librerías y creación del dataset


```python
import pandas as pd
import numpy as np

# Dataset con valores nulos y duplicados
datos = {
    "Nombre": ["Ana", "Luis", "Pedro", "María", "Luis", "Carlos", "Ana"],
    "Edad": [25, np.nan, 28, 22, np.nan, 35, 25],
    "Departamento": ["Ventas", "IT", "Ventas", None, "IT", "IT", "Ventas"],
    "Salario": [1500, 1800, np.nan, 1400, 1800, 2200, 1500]
}

df = pd.DataFrame(datos)
df
```

## 1. Identificación de valores nulos


```python
# Ver valores nulos por columna
df.isnull().sum()
# Mostrar filas con al menos un valor nulo
df[df.isnull().any(axis=1)]
```

## 2. Estrategias de imputación de valores nulos
### Estrategia 1: Imputación con la media (columna numérica)


```python
# Reemplazar valores nulos en Edad con la media
media_edad = df["Edad"].mean()
df["Edad_media"] = df["Edad"].fillna(media_edad)

df[["Edad", "Edad_media"]]
```

### Estrategia 2: Imputación con la mediana


```python
# Reemplazar valores nulos en Salario con la mediana
mediana_salario = df["Salario"].median()
df["Salario_mediana"] = df["Salario"].fillna(mediana_salario)

df[["Salario", "Salario_mediana"]]
```

### Estrategia 3: Imputación con un valor fijo


```python
# Reemplazar valores nulos en Departamento con un valor fijo
df["Departamento_fijo"] = df["Departamento"].fillna("No especificado")

df[["Departamento", "Departamento_fijo"]]
```

## 3. Detección y eliminación de duplicados


```python
# Detectar filas duplicadas
df.duplicated()
# Mostrar filas duplicadas
df[df.duplicated()]
# Eliminar duplicados
df_sin_duplicados = df.drop_duplicates()
df_sin_duplicados
```
