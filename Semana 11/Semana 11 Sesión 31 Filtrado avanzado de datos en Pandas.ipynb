En este notebook se presentan ejemplos de filtrado avanzado en Pandas utilizando
condiciones múltiples, operadores lógicos, métodos como `isin`, `between`, `query`
y filtrado por texto.

## Importación de librerías y creación del DataFrame base


```python
import pandas as pd
import numpy as np

# Dataset de ejemplo
datos = {
    "Nombre": ["Ana", "Luis", "Pedro", "María", "Carlos", "Lucía", "Jorge", "Sofía", "Miguel", "Elena"],
    "Edad": [25, 30, 28, 22, 35, 27, 40, 24, 33, 29],
    "Departamento": ["Ventas", "IT", "Ventas", "Marketing", "IT",
                     "RRHH", "Finanzas", "Marketing", "IT", "Ventas"],
    "Salario": [1500, 1800, 1700, 1400, 2200, 1600, 2500, 1450, 2100, 1750]
}

df = pd.DataFrame(datos)
df
```

## Ejemplo 1: Filtrado con múltiples condiciones (AND)


```python
# Empleados mayores de 30 años y salario mayor a 2000
filtro_1 = df[(df["Edad"] > 30) & (df["Salario"] > 2000)]
filtro_1
```

## Ejemplo 2: Filtrado con condición OR


```python
# Empleados del departamento IT o Ventas
filtro_2 = df[(df["Departamento"] == "IT") | (df["Departamento"] == "Ventas")]
filtro_2
```

## Ejemplo 3: Filtrado usando isin()


```python
# Empleados que pertenecen a IT o Finanzas
filtro_3 = df[df["Departamento"].isin(["IT", "Finanzas"])]
filtro_3
```

## Ejemplo 4: Filtrado por rango con between()


```python
# Empleados con edad entre 25 y 35 años
filtro_4 = df[df["Edad"].between(25, 35)]
filtro_4
```

## Ejemplo 5: Filtrado usando query()


```python
# Empleados con salario mayor a 1700 y edad menor a 35
filtro_5 = df.query("Salario > 1700 and Edad < 35")
filtro_5
```

## Ejemplo 6: Filtrado por texto (str.contains)


```python
# Empleados cuyo nombre contiene la letra 'a' (mayúscula o minúscula)
filtro_6 = df[df["Nombre"].str.contains("a", case=False)]
filtro_6
```
