En este notebook se presentan ejemplos prácticos de transformación de datos usando Pandas,
incluyendo la creación de nuevas columnas, el uso de `apply()` con funciones lambda
y el reemplazo de valores con `map()` y `replace()`.

## Importación de librerías y creación del DataFrame base


```python
import pandas as pd
```

```python
# DataFrame de ejemplo
datos = {
    "Nombre": ["Ana", "Luis", "Pedro", "María", "Carlos"],
    "Edad": [25, 30, 28, 22, 35],
    "Departamento": ["Ventas", "IT", "Ventas", "Marketing", "IT"],
    "Salario": [1500, 1800, 1700, 1400, 2200]
}

df = pd.DataFrame(datos)
df
```

## Ejemplo 1: Creación de nuevas columnas


```python
# Crear una nueva columna con salario anual
df["Salario_Anual"] = df["Salario"] * 12

# Crear una columna booleana
df["Es_Mayor_30"] = df["Edad"] > 30

df
```

## Ejemplo 2: Transformaciones con apply() y lambda


```python
# Clasificar empleados según su salario
df["Nivel_Salarial"] = df["Salario"].apply(
    lambda x: "Alto" if x >= 2000 else "Medio" if x >= 1600 else "Bajo"
)

df
```

```python
# Transformar nombres a mayúsculas
df["Nombre_Mayusculas"] = df["Nombre"].apply(lambda x: x.upper())
df
```

## Ejemplo 3: Reemplazo de valores con map() y replace()


```python
# Reemplazar nombres de departamentos con abreviaturas usando map()
map_departamentos = {
    "Ventas": "VTA",
    "IT": "TEC",
    "Marketing": "MKT"
}

df["Depto_Abrev"] = df["Departamento"].map(map_departamentos)
df
```

```python
# Reemplazar valores directamente con replace()
df["Departamento_Num"] = df["Departamento"].replace({
    "Ventas": 1,
    "IT": 2,
    "Marketing": 3
})

df
```
