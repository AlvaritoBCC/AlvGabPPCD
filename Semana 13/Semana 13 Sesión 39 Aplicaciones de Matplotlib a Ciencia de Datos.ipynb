## Importación de librerías


```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Carga de un dataset real (Iris)

Se utiliza el dataset Iris, ampliamente usado en ciencia de datos.


```python
# Cargar dataset Iris
try:
    from sklearn.datasets import load_iris
    iris = load_iris(as_frame=True)
    df = iris.frame
except ImportError:
    raise ImportError("Se requiere scikit-learn para cargar el dataset Iris")

df.head()
```

## Exploración básica del dataset


```python
df.info()
df.describe()
```

## Histograma: Distribución del largo del sépalo
Este histograma muestra la distribución del largo del sépalo en el dataset Iris.


```python
plt.figure()
plt.hist(df["sepal length (cm)"], bins=20)
plt.title("Histograma del largo del sépalo")
plt.xlabel("Largo del sépalo (cm)")
plt.ylabel("Frecuencia")
plt.savefig("histograma_sepal_length.png")
plt.show()
```

## Diagrama de dispersión: Largo vs ancho del sépalo
El gráfico permite observar la relación entre el largo y el ancho del sépalo.


```python
plt.figure()
plt.scatter(df["sepal length (cm)"], df["sepal width (cm)"])
plt.title("Diagrama de dispersión: sépalo")
plt.xlabel("Largo del sépalo (cm)")
plt.ylabel("Ancho del sépalo (cm)")
plt.savefig("dispersion_sepal.png")
plt.show()
```

## Boxplot: Distribución del ancho del pétalo
El boxplot permite identificar la mediana, la dispersión y posibles valores atípicos.


```python
plt.figure()
plt.boxplot(df["petal width (cm)"])
plt.title("Boxplot del ancho del pétalo")
plt.ylabel("Ancho del pétalo (cm)")
plt.savefig("boxplot_petal_width.png")
plt.show()
```
