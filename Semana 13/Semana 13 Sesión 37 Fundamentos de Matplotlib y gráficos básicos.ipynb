## Importación de librerías


```python
import matplotlib.pyplot as plt
import numpy as np
```

## Gráfico 1: Gráfico de líneas
Este gráfico representa la relación cuadrática entre X e Y.


```python
x = np.linspace(0, 10, 50)
y = x ** 2

plt.figure()
plt.plot(x, y)
plt.title("Gráfico de líneas: y = x²")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.show()
```

## Gráfico 2: Gráfico de barras
El gráfico de barras permite comparar valores entre distintas categorías.


```python
categorias = ["A", "B", "C", "D"]
valores = [15, 30, 10, 25]

plt.figure()
plt.bar(categorias, valores)
plt.title("Gráfico de barras")
plt.xlabel("Categorías")
plt.ylabel("Valores")
plt.show()
```

## Gráfico 3: Gráfico de dispersión
Este gráfico muestra la relación entre dos variables aleatorias.


```python
np.random.seed(1)
x = np.random.rand(40)
y = np.random.rand(40)

plt.figure()
plt.scatter(x, y)
plt.title("Gráfico de dispersión")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.show()
```
