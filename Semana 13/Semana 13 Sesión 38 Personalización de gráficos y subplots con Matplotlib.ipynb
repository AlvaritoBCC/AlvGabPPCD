En este notebook se presentan ejemplos de gráficos personalizados utilizando Matplotlib,
incluyendo colores, estilos de línea, marcadores, leyendas claras y el uso de subplots
con un título general.


## Importación de librerías


```python
import matplotlib.pyplot as plt
import numpy as np
```

## Ejemplo 1: Gráfico de líneas personalizado
Este gráfico muestra las funciones seno y coseno con diferentes estilos de línea y marcadores.


```python
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure()
plt.plot(x, y1, linestyle='-', marker='o', label='Seno')
plt.plot(x, y2, linestyle='--', marker='s', label='Coseno')
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.title("Funciones trigonométricas")
plt.legend()
plt.show()
```

## Ejemplo 2: Gráfico de dispersión personalizado
El diagrama de dispersión utiliza marcadores personalizados y una leyenda descriptiva.

```python
np.random.seed(0)
x = np.random.rand(30)
y = np.random.rand(30)

plt.figure()
plt.scatter(x, y, marker='^', label='Puntos aleatorios')
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.title("Gráfico de dispersión personalizado")
plt.legend()
plt.show()
```

## Ejemplo 3: Subplots con título general (suptitle)
En esta figura se presentan dos subplots con un título general que describe el conjunto
de gráficos.


```python
x = np.linspace(0, 2 * np.pi, 100)

fig, axs = plt.subplots(2, 1)
fig.suptitle("Subplots de funciones trigonométricas")

axs[0].plot(x, np.sin(x), color='blue', label='Seno')
axs[0].set_title("Seno")
axs[0].legend()

axs[1].plot(x, np.cos(x), color='red', linestyle='--', label='Coseno')
axs[1].set_title("Coseno")
axs[1].legend()

plt.show()
```
