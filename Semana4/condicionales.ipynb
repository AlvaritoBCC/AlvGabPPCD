# Contenidos

Sintaxis de if, elif, else


```python
if condicion:
    # bloque de código si la condición es verdadera
elif otra_condicion:
    # bloque de código si la segunda condición es verdadera
else:
    # bloque de código si ninguna condición anterior es verdadera
```

| Operador | Significado       | Ejemplo   | Resultado |
| -------- | ----------------- | --------- | --------- |
| `==`     | Igual a           | `5 == 5`  | `True`    |
| `!=`     | Diferente de      | `5 != 3`  | `True`    |
| `<`      | Menor que         | `4 < 7`   | `True`    |
| `>`      | Mayor que         | `10 > 15` | `False`   |
| `<=`     | Menor o igual que | `3 <= 3`  | `True`    |
| `>=`     | Mayor o igual que | `8 >= 5`  | `True`    |

### Condicionales anidados

```python
edad = int(input("Ingresa tu edad: "))

if edad >= 18:
    print("Eres mayor de edad.")
    if edad >= 60:
        print("Además, eres adulto mayor.")
else:
    print("Eres menor de edad.")
```

### Usando los condicionales

```python
# Clasificador de temperaturas
temperaturas = [10, 20, 28, 33, 41]

for t in temperaturas:
    if t < 15:
        print(f"{t}°C -> Frío")
    elif 15 <= t < 30:
        print(f"{t}°C -> Agradable")
    elif 30 <= t < 40:
        print(f"{t}°C -> Caluroso")
    else:
        print(f"{t}°C -> Extremo")
```

```python
# Detectar números misteriosos
numeros = [17, 27, 33, 57, 70, 87, 99]

print("🔎 Números misteriosos encontrados:")
for n in numeros:
    if n % 3 == 0 and str(n).endswith("7"):
        print(f" {n} es misterioso")
    else:
        print(f" {n} no cumple las condiciones")
```

```python
# Máquina de cambio
cambio = int(input(" Ingresa el monto a devolver: "))
monedas = [10, 5, 1]

for m in monedas:
    if cambio >= m:
        cantidad = cambio // m  # Cuántas monedas de ese tipo
        cambio %= m  # Actualizar el cambio restante
        print(f"{cantidad} moneda(s) de ${m}")
    elif cambio == 0:
        print(" No queda más cambio por entregar.")
        break
```
