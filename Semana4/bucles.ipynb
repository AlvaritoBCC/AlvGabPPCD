## Contenidos
- Sintaxis de for con listas, cadenas y range()

```python
# Recorriendo una lista
for elemento in [10, 20, 30]:
    print(elemento)

# Recorriendo una cadena de texto
for caracter in "Python":
    print(caracter)

# Recorriendo una secuencia numérica
for i in range(5):
    print(i)
```

```python
# Estructura básica
while condicion:
    # bloque de código
```

```python
contador = 0
while contador < 5:
    print("Contador:", contador)
    contador += 1
```

| Instrucción | Descripción                                                      |
| ----------- | ---------------------------------------------------------------- |
| `break`     | Interrumpe completamente el bucle.                               |
| `continue`  | Salta a la siguiente iteración sin ejecutar el resto del bloque. |
| `pass`      | No hace nada (se usa como marcador o código temporal).           |

# Ejercicios Resueltos con for y while

```python
n = int(input("Ingresa un número: "))
suma = 0

for i in range(1, n + 1):
    suma += i

print(f"La sumatoria de los primeros {n} números es {suma}")
```

```python
n = int(input("Ingresa un número: "))
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(f"El factorial de {n} es {factorial}")
```

```python
palabra = input("Escribe una palabra: ").lower()
vocales = "aeiou"
contador = 0

for letra in palabra:
    if letra in vocales:
        contador += 1

print(f"La palabra '{palabra}' tiene {contador} vocal(es).")
```

```python
cuadrados = []

for i in range(1, 11):
    cuadrados.append(i ** 2)

print("Lista de cuadrados:", cuadrados)
```

```python
contraseña_correcta = "1234"
intentos = 3

while intentos > 0:
    clave = input("Ingresa tu contraseña: ")

    if clave == contraseña_correcta:
        print("✅ Acceso concedido. Bienvenido.")
        break
    else:
        intentos -= 1
        print(f" Contraseña incorrecta. Intentos restantes: {intentos}")

if intentos == 0:
    print(" Tarjeta bloqueada por demasiados intentos fallidos.")
```
