```python
# Validador de Contraseña con Métricas

contraseña = input("🔐 Ingresa una contraseña: ")

letras = sum(1 for c in contraseña if c.isalpha())
numeros = sum(1 for c in contraseña if c.isdigit())

if len(contraseña) < 8:
    print(" La contraseña debe tener al menos 8 caracteres.")
elif letras == 0 or numeros == 0:
    print(" Debe contener al menos una letra y un número.")
else:
    print(" Contraseña válida.")
    print(f" Longitud: {len(contraseña)} | Letras: {letras} | Números: {numeros}")
```

```python
# Secuencia de Fibonacci con conteo de pares e impares

n = int(input(" Ingresa la cantidad de términos de Fibonacci: "))

a, b = 0, 1
pares, impares = 0, 0

print(" Secuencia de Fibonacci:")

for i in range(n):
    print(a, end=" ")
    if a % 2 == 0:
        pares += 1
    else:
        impares += 1
    a, b = b, a + b

print(f"\n Pares: {pares} | Impares: {impares}")
```

```python
import random

# Juego: Piedra, Papel o Tijera (Mejor de 3)

opciones = ["piedra", "papel", "tijera"]
usuario_puntos = 0
pc_puntos = 0

while usuario_puntos < 2 and pc_puntos < 2:
    usuario = input("\n✊✋✌️ Elige (piedra, papel o tijera): ").lower()
    pc = random.choice(opciones)
    print(f" La computadora eligió: {pc}")

    if usuario not in opciones:
        print(" Opción inválida. Intenta de nuevo.")
        continue

    if usuario == pc:
        print(" Empate.")
    elif (usuario == "piedra" and pc == "tijera") or \
         (usuario == "papel" and pc == "piedra") or \
         (usuario == "tijera" and pc == "papel"):
        print(" ¡Ganaste esta ronda!")
        usuario_puntos += 1
    else:
        print(" La computadora gana esta ronda.")
        pc_puntos += 1

    print(f"Marcador: Tú {usuario_puntos} - {pc_puntos} PC")

# Resultado final
if usuario_puntos > pc_puntos:
    print("\n ¡Felicidades, ganaste el juego!")
else:
    print("\n La computadora ganó esta vez.")
```
