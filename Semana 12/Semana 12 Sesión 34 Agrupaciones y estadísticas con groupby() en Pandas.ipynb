En este notebook se genera un dataset falso de reproducciones musicales utilizando Faker,
se guarda en un archivo CSV y se realizan análisis estadísticos usando `groupby()`.

El generador permite crear un número variable de registros (n), que para fines prácticos
se fija en 100,000.


## Importación de librerías


```python
import pandas as pd
import numpy as np
from faker import Faker
```

## Configuración inicial y parámetros


```python
# Semillas para reproducibilidad
np.random.seed(42)
fake = Faker()
Faker.seed(42)

# Número de registros (puede cambiarse)
n = 100_000

# Número de canciones distintas
num_canciones = 100
```

## Generación de canciones únicas


```python
# Crear lista de canciones únicas
canciones = [fake.sentence(nb_words=3).replace(".", "") for _ in range(num_canciones)]
canciones
```

## Generación del dataset fake


```python
datos = {
    "Cancion": np.random.choice(canciones, n),
    "Año": np.random.randint(2018, 2025, n),
    "Mes": np.random.randint(1, 13, n),
    "Reproducciones": np.random.randint(1, 5000, n),
    "Likes": np.random.randint(0, 3000, n)
}

df = pd.DataFrame(datos)
df.head()
```

## Guardado del dataset en CSV


```python
df.to_csv("fake_canciones.csv", index=False)
```

## Exploración inicial del dataset


```python
df.shape
df.info()
df.describe()
```

## Agrupaciones y estadísticas con groupby()

### 1. Reproducciones totales por canción



```python
repro_por_cancion = df.groupby("Cancion")["Reproducciones"].sum().sort_values(ascending=False)
repro_por_cancion.head(10)
```

### 2. Likes promedio por canción


```python
likes_promedio = df.groupby("Cancion")["Likes"].mean().sort_values(ascending=False)
likes_promedio.head(10)
```

### 3. Reproducciones por año


```python
repro_por_anio = df.groupby("Año")["Reproducciones"].sum()
repro_por_anio
```

### 4. Reproducciones por mes


```python
repro_por_mes = df.groupby("Mes")["Reproducciones"].sum()
repro_por_mes
```

### 5. Análisis combinado: Año y canción


```python
repro_anio_cancion = df.groupby(["Año", "Cancion"])["Reproducciones"].sum()
repro_anio_cancion.head(10)
```
