En este notebook se utilizan tablas dinámicas (`pivot_table`) para analizar datos de ventas
desde distintos enfoques: región, mes, producto y categoría.

## Importación de librerías


```python
import pandas as pd
import numpy as np
```

## DataFrame base (contexto: ventas)


```python
# Dataset base de ventas
data = {
    "Region": ["Norte", "Sur", "Este", "Oeste"] * 6,
    "Mes": ["Enero", "Febrero", "Marzo"] * 8,
    "Producto": ["Laptop", "Mouse", "Teclado"] * 8,
    "Categoria": ["Tecnología", "Accesorios", "Accesorios"] * 8,
    "Ventas": np.random.randint(100, 5000, 24)
}

df = pd.DataFrame(data)
df
```

## Tabla dinámica 1: Ventas totales por región y mes


```python
pivot_region_mes = pd.pivot_table(
    df,
    values="Ventas",
    index="Region",
    columns="Mes",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

pivot_region_mes
```

## Tabla dinámica 2: Promedio de ventas por producto y región


```python
pivot_producto_region = pd.pivot_table(
    df,
    values="Ventas",
    index="Producto",
    columns="Region",
    aggfunc="mean",
    fill_value=0
)

pivot_producto_region
```

## Tabla dinámica 3: Múltiples funciones de agregación por categoría y mes


```python
pivot_categoria_mes = pd.pivot_table(
    df,
    values="Ventas",
    index="Categoria",
    columns="Mes",
    aggfunc=["sum", "mean", "max"],
    fill_value=0
)

pivot_categoria_mes
```
