En este notebook se presentan ejemplos prácticos de combinación de DataFrames usando
`merge()`, `concat()` y `join()`, aplicados a contextos reales como estudiantes,
cursos, ventas y productos.

## Importación de librerías


```python
import pandas as pd
```

## DataFrames base (contexto: estudiantes y cursos)


```python
# DataFrame de estudiantes
estudiantes = pd.DataFrame({
    "id_estudiante": [1, 2, 3, 4],
    "Nombre": ["Ana", "Luis", "Pedro", "María"],
    "Carrera": ["Ingeniería", "Medicina", "Derecho", "Ingeniería"]
})

# DataFrame de cursos
cursos = pd.DataFrame({
    "id_estudiante": [1, 2, 4, 5],
    "Curso": ["Python", "Biología", "Estadística", "Historia"]
})

estudiantes, cursos
```

## 1. Uso de merge() con distintos tipos de join

### Inner join
Solo se conservan los registros que existen en ambos DataFrames.
Solo aparecen los estudiantes que tienen curso asignado en ambos DataFrames.


```python
merge_inner = pd.merge(estudiantes, cursos, on="id_estudiante", how="inner")
merge_inner
```

### Right join
Se conservan todos los registros del DataFrame derecho.


```python
merge_right = pd.merge(estudiantes, cursos, on="id_estudiante", how="right")
merge_right
```

### Outer join
Se conservan todos los registros de ambos DataFrames.


```python
merge_outer = pd.merge(estudiantes, cursos, on="id_estudiante", how="outer")
merge_outer
```

## 2. Uso de pd.concat()

### Concatenación vertical (mismas columnas)
Contexto: ventas de dos meses distintos.


```python
ventas_enero = pd.DataFrame({
    "Producto": ["Laptop", "Mouse", "Teclado"],
    "Ventas": [10, 25, 15]
})

ventas_febrero = pd.DataFrame({
    "Producto": ["Laptop", "Mouse", "Monitor"],
    "Ventas": [8, 30, 12]
})

ventas_vertical = pd.concat([ventas_enero, ventas_febrero], axis=0)
ventas_vertical
```

### Concatenación horizontal (más columnas)


```python
info_producto = pd.DataFrame({
    "Precio": [1000, 20, 45]
})

ventas_horizontal = pd.concat([ventas_enero, info_producto], axis=1)
ventas_horizontal
```

## 3. Uso de join() con índices

### Join usando índices simples


```python
# DataFrames con índice
productos = pd.DataFrame({
    "Producto": ["Laptop", "Mouse", "Teclado"],
    "Stock": [5, 50, 20]
}).set_index("Producto")

precios = pd.DataFrame({
    "Producto": ["Laptop", "Mouse", "Monitor"],
    "Precio": [1000, 20, 300]
}).set_index("Producto")

join_simple = productos.join(precios)
join_simple
```

### Join convirtiendo una columna en índice con set_index()


```python
# Convertir columna a índice antes del join
estudiantes_index = estudiantes.set_index("id_estudiante")
cursos_index = cursos.set_index("id_estudiante")

join_index = estudiantes_index.join(cursos_index)
join_index
```
