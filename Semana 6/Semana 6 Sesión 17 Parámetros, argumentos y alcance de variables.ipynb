## ¿Qué son los parámetros con valores por defecto?

Los parámetros con valores por defecto permiten que una función sea llamada sin proporcionar todos los argumentos.
Si el argumento no se pasa, se utiliza el valor definido por defecto.


```python

def potencia(base, exponente=2):
    """
    Calcula la potencia de un número.

    Parámetros:
    base (int o float): número base.
    exponente (int): exponente, por defecto es 2.

    Retorna:
    int o float: resultado de elevar base al exponente.

    Ejemplo:
    potencia(3) -> 9
    potencia(2, 3) -> 8
    """
    return base ** exponente

print(potencia(4))
print(potencia(2, 3))
assert potencia(3) == 9
assert potencia(2, 4) == 16
```

```python

def precio_con_iva(precio, iva=0.16):
    """
    Calcula el precio final con IVA.

    Parámetros:
    precio (float): precio base del producto.
    iva (float): porcentaje de IVA, por defecto 0.16.

    Retorna:
    float: precio final con IVA.

    Ejemplo:
    precio_con_iva(100) -> 116.0
    """
    return precio * (1 + iva)

print(precio_con_iva(100))
assert precio_con_iva(100) == 116.0
```

```python

def convertir_temp(valor, a='F'):
    """
    Convierte temperaturas entre Celsius y Fahrenheit.

    Parámetros:
    valor (float): temperatura a convertir.
    a (str): 'F' para Fahrenheit o 'C' para Celsius.

    Retorna:
    float: temperatura convertida.

    Ejemplo:
    convertir_temp(0) -> 32.0
    convertir_temp(32, 'C') -> 0.0
    """
    if a == 'F':
        return (valor * 9/5) + 32
    elif a == 'C':
        return (valor - 32) * 5/9
    else:
        raise ValueError("Parámetro 'a' debe ser 'F' o 'C'")

print(convertir_temp(0))
print(convertir_temp(32, 'C'))
assert convertir_temp(0) == 32.0
```

```python

def descuento(precio, porcentaje=0.10):
    """
    Calcula el precio con descuento.

    Parámetros:
    precio (float): precio original.
    porcentaje (float): porcentaje de descuento, por defecto 0.10.

    Retorna:
    float: precio con descuento.

    Ejemplo:
    descuento(100) -> 90.0
    """
    return precio * (1 - porcentaje)

print(descuento(200))
assert descuento(100) == 90.0
```

```python

def formatear_nombre(nombre, apellido, mayus=False):
    """
    Formatea un nombre completo.

    Parámetros:
    nombre (str): nombre de la persona.
    apellido (str): apellido de la persona.
    mayus (bool): si es True, convierte a mayúsculas.

    Retorna:
    str: nombre formateado.

    Ejemplo:
    formatear_nombre("Juan", "Perez") -> "Juan Perez"
    """
    nombre_completo = f"{nombre} {apellido}"
    return nombre_completo.upper() if mayus else nombre_completo

print(formatear_nombre("Juan", "Perez"))
print(formatear_nombre("Juan", "Perez", True))
assert formatear_nombre("Ana", "Lopez") == "Ana Lopez"
```
