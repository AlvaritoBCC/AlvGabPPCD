```python

def cuadrado(n):
    """
    Calcula el cuadrado de un número.
    
    Ejemplo:
    cuadrado(4) -> 16
    """
    return n * n

# Pruebas
print(cuadrado(5))
assert cuadrado(3) == 9
```

```python

def contar_vocales(texto):
    """
    Cuenta el número de vocales en una cadena.
    
    Ejemplo:
    contar_vocales("hola") -> 2
    """
    vocales = "aeiouAEIOU"
    return sum(1 for c in texto if c in vocales)

# Pruebas
print(contar_vocales("Python"))
assert contar_vocales("hola") == 2
```

```python

def maximo_lista(nums):
    """
    Devuelve el valor máximo de una lista de números.
    
    Ejemplo:
    maximo_lista([1, 5, 3]) -> 5
    """
    return max(nums)

# Pruebas
print(maximo_lista([10, 3, 7]))
assert maximo_lista([1, 2, 3]) == 3
```

```python

def es_par(n):
    """
    Determina si un número es par.
    
    Ejemplo:
    es_par(4) -> True
    """
    return n % 2 == 0

# Pruebas
print(es_par(10))
assert es_par(3) is False
```

```python

def promedio_lista(nums):
    """
    Calcula el promedio de una lista de números.
    
    Ejemplo:
    promedio_lista([2, 4, 6]) -> 4.0
    """
    return sum(nums) / len(nums)

# Pruebas
print(promedio_lista([5, 10, 15]))
assert promedio_lista([2, 4, 6]) == 4
```
