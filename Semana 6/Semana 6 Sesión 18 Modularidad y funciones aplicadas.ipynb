Este notebook define funciones reutilizables para limpiar datos, las prueba con un dataset CSV pequeño y muestra estadísticas básicas.

```python
import pandas as pd
import numpy as np
from statistics import mode
```

## 1. Definición de funciones de limpieza

```python
def eliminar_duplicados(df):
    """
    Elimina filas duplicadas de un DataFrame.
    
    Parámetros:
    df (pd.DataFrame): DataFrame de entrada.
    
    Retorna:
    pd.DataFrame: DataFrame sin duplicados.
    """
    return df.drop_duplicates()

def reemplazar_nulos(df, valor=0):
    """
    Reemplaza valores nulos en un DataFrame.
    
    Parámetros:
    df (pd.DataFrame): DataFrame de entrada.
    valor: Valor con el que se reemplazarán los nulos.
    
    Retorna:
    pd.DataFrame: DataFrame sin valores nulos.
    """
    return df.fillna(valor)

def convertir_tipos(df, columnas):
    """
    Convierte columnas específicas a tipo numérico.
    
    Parámetros:
    df (pd.DataFrame): DataFrame de entrada.
    columnas (list): Lista de columnas a convertir.
    
    Retorna:
    pd.DataFrame: DataFrame con columnas convertidas.
    """
    for col in columnas:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    return df
```

## 2. Creación y carga de un dataset CSV

```python
# Crear un dataset pequeño
data = {
    'nombre': ['Ana', 'Luis', 'Ana', 'Pedro'],
    'edad': ['25', '30', '25', None],
    'salario': [3000, 3500, 3000, 4000]
}

df_original = pd.DataFrame(data)
df_original.to_csv('datos_ejemplo.csv', index=False)

# Cargar el CSV
df = pd.read_csv('datos_ejemplo.csv')
df
```

## 3. Aplicación de las funciones de limpieza

```python
df = eliminar_duplicados(df)
df = convertir_tipos(df, ['edad'])
df = reemplazar_nulos(df, valor=df['edad'].mean())
df
```

## 4. Estadísticas descriptivas

```python
columna = 'edad'

media = df[columna].mean()
mediana = df[columna].median()
moda = mode(df[columna])
rango = df[columna].max() - df[columna].min()

print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Moda: {moda}")
print(f"Rango: {rango}")
```
