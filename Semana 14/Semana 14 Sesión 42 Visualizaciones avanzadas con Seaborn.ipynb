En este notebook se realiza un análisis exploratorio visual completo utilizando Seaborn

patrones, relaciones y posibles valores atípicos.


## Importación de librerías


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Carga del dataset

Se utiliza el dataset **penguins**, incluido en Seaborn, que contiene información
biométrica de diferentes especies de pingüinos.


```python
df = sns.load_dataset("penguins")
df.head()
```

## Limpieza básica de datos


```python
# Eliminación de valores nulos para el análisis
df = df.dropna()
df.info()
```

## Configuración de estilo general


```python
sns.set_style("whitegrid")
```

## Visualización 1: Histograma – Distribución del peso corporal
 
La distribución del peso corporal presenta una forma aproximadamente normal, con
diferencias notables entre individuos más ligeros y más pesados.


```python
plt.figure()
sns.histplot(df["body_mass_g"], bins=20, kde=True)
plt.title("Distribución del peso corporal de los pingüinos")
plt.xlabel("Peso corporal (g)")
plt.ylabel("Frecuencia")
plt.show()
```

## Visualización 2: Boxplot – Peso corporal por especie

Se observan diferencias claras en el peso corporal entre especies, así como algunos
posibles valores atípicos.


```python
plt.figure()
sns.boxplot(data=df, x="species", y="body_mass_g")
plt.title("Peso corporal por especie")
plt.xlabel("Especie")
plt.ylabel("Peso corporal (g)")
plt.show()
```

## Visualización 3: Scatterplot – Largo vs profundidad del pico

Existe una separación visible entre especies, lo que sugiere que estas variables
pueden ser útiles para clasificación.


```python
plt.figure()
sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species"
)
plt.title("Relación entre largo y profundidad del pico")
plt.xlabel("Largo del pico (mm)")
plt.ylabel("Profundidad del pico (mm)")
plt.show()
```

## Visualización 4: Heatmap de correlación

Se observa una fuerte correlación entre el peso corporal y las dimensiones del pico
y las aletas, indicando relaciones biológicas consistentes.


```python
plt.figure()
correlacion = df.select_dtypes(include="number").corr()
sns.heatmap(correlacion, annot=True, cmap="coolwarm")
plt.title("Mapa de calor de correlaciones")
plt.show()
```

## Visualización 5 (Avanzada): Pairplot por especie

El pairplot permite analizar simultáneamente múltiples relaciones entre variables.
Se observa una clara separación entre especies en varias combinaciones de variables.


```python
sns.pairplot(
    df,
    hue="species",
    diag_kind="kde"
)
```

## Visualización 6 (Avanzada): lmplot – Relación con regresión
 
Existe una relación lineal positiva entre el largo de las aletas y el peso corporal,
con pendientes ligeramente distintas según la especie.


```python
sns.lmplot(
    data=df,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)
```

Este mini-EDA visual demuestra cómo Seaborn permite explorar datos de forma clara
y efectiva. Las visualizaciones avanzadas facilitan la identificación de patrones,
relaciones y diferencias entre grupos, siendo una base sólida para análisis más
profundos o modelos de machine learning.
