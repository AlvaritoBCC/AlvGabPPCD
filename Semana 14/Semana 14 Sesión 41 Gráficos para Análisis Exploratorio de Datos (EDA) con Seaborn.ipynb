En este notebook se realiza un análisis exploratorio de datos (EDA) utilizando Seaborn
sobre un dataset real. Se incluyen visualizaciones clave para identificar patrones,
tendencias y posibles valores atípicos.

## Importación de librerías


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Carga de dataset real

Se utiliza el dataset `tips`, incluido en Seaborn, que contiene información sobre
consumo y propinas en un restaurante.


```python
df = sns.load_dataset("tips")
df.head()
```

## Configuración de estilo


```python
sns.set_style("whitegrid")
```

## 1. Scatterplot: Relación entre total de la cuenta y propina

Existe una relación positiva entre el total de la cuenta y la propina. Las cuentas
más altas tienden a generar propinas mayores. No se observan outliers extremos,
aunque algunas cuentas grandes destacan.

```python
plt.figure()
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time"
)
plt.title("Relación entre total de la cuenta y propina")
plt.xlabel("Total de la cuenta")
plt.ylabel("Propina")
plt.show()
```

## 2. Boxplot: Distribución de propinas por día
Se observan diferencias en la distribución de propinas entre días. Algunos días
presentan mayor variabilidad y posibles valores atípicos (outliers).

```python
plt.figure()
sns.boxplot(
    data=df,
    x="day",
    y="tip"
)
plt.title("Distribución de propinas por día")
plt.xlabel("Día")
plt.ylabel("Propina")
plt.show()
```

## 3. Heatmap de correlación entre variables numéricas

La mayor correlación se presenta entre el total de la cuenta y la propina. Otras
variables numéricas muestran correlaciones bajas o moderadas.

```python
plt.figure()
correlacion = df.select_dtypes(include="number").corr()
sns.heatmap(correlacion, annot=True, cmap="coolwarm")
plt.title("Mapa de calor de correlaciones")
plt.show()
```
