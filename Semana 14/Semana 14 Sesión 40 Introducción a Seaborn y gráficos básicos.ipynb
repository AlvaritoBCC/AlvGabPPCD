## Importación de librerías


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Carga de un dataset real

Se utiliza el dataset `tips`, incluido en Seaborn, que contiene información sobre cuentas
en un restaurante.


```python
df = sns.load_dataset("tips")
df.head()
```

## Configuración de estilo general


```python
sns.set_style("whitegrid")
```

## Gráfico 1: Histplot – Distribución del total de la cuenta
Este histograma permite analizar cómo se distribuyen los montos totales de las cuentas.


```python
plt.figure()
sns.histplot(df["total_bill"], bins=20, kde=True)
plt.title("Distribución del total de la cuenta")
plt.xlabel("Total de la cuenta")
plt.ylabel("Frecuencia")
plt.show()
```

## Gráfico 2: Countplot – Frecuencia por día
Este gráfico muestra cuántos registros existen para cada día de la semana.


```python
plt.figure()
sns.countplot(data=df, x="day")
plt.title("Frecuencia de visitas por día")
plt.xlabel("Día")
plt.ylabel("Cantidad de registros")
plt.show()
```

## Gráfico 3: Barplot – Promedio de propina por día
El barplot representa el promedio de propina por día, permitiendo comparar el
comportamiento de los clientes a lo largo de la semana.


```python
plt.figure()
sns.histplot(df["total_bill"], bins=20, kde=True)
plt.title("Distribución del total de la cuenta")
plt.xlabel("Total de la cuenta")
plt.ylabel("Frecuencia")
plt.show()
```
